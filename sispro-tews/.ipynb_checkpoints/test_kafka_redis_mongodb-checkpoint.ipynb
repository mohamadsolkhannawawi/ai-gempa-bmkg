## Test Kafka

Running consumer dan producer di 2 instance yang berbeda (karena consumer akan forever loop) <br />

Bila berhasil, consumer akan memberikan output sesuai yang diberikan producer <br />

Bila dalam error message ada masalah api version, bisa ditambahkan argument api_version=(0, 10, 0) ketika inisialisasi kafka consumer / producer. Error ini seharusnya tidak terjadi karena api version seharusnya diambil dari broker. Kemungkinan error terjadi karena koneksi ke broker terputus ketika sedang mengambil api version dari broker

### Consumer

```python
from kafka import KafkaConsumer

consumer = KafkaConsumer(
    'test',
    bootstrap_servers='152.118.31.54:9999'
)

for message in consumer:
    print(f"Received message: {message.value}")
```

### Producer

```python
import json
from kafka import KafkaProducer

producer = KafkaProducer(bootstrap_servers=["152.118.31.54:9999"])
producer.send("test", "test".encode())
producer.flush()
```

## Test Redis

Bila berhasil akan mengeluarkan output b'test'

```python
import redis

redis_host = "152.118.31.54"
redis_port = "6380"
# Connect to Redis
redis_client = redis.Redis(host=redis_host, port=int(redis_port), db=0)

redis_client.set("test", "Test")
print(redis_client.get("test"))
```
